```python
import boto3
from datetime import date
import json

client = boto3.client("ce")

response = client.get_cost_and_usage(
    TimePeriod={
        "Start": "2025-08-30",   # inclusive
        "End": "2025-09-09"      # exclusive
    },
    Granularity="DAILY",
    Metrics=["UnblendedCost"],
    GroupBy=[
        {
            "Type": "DIMENSION",
            "Key": "SERVICE"
        }
    ],
    Filter={
        "Dimensions": {
            "Key": "LINKED_ACCOUNT",
            "Values": [
                "014498663963"
            ]
        }
    }
)

```

```python
ResultsByTime = response.get('ResultsByTime', [])
ResultsByTime
```

```python
a = ResultsByTime[9].get('Groups', [])
```

```python
clm = 0

for i in a:
    c = i.get('Metrics').get('UnblendedCost').get('Amount')
    c = float(c)
    print(i.get('Keys')[0], c)
    clm += c
print("Total:", clm)
```
